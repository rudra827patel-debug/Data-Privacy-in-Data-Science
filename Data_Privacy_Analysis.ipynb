```python
print("Number of records:", len(df))
print("Number of columns:", len(df.columns))
```

```python
personal_columns = [
    "Name",
    "Age",
    "Gender",
    "City",
    "Email"
]

sensitive_columns = [
    "Payment_Method"
]

print("Personal Information:")
print(personal_columns)

print("\nSensitive Information:")
print(sensitive_columns)
```

```python
def mask_email(email):
    first_letter = email[0]
    domain = email[email.index("@"):]
    return first_letter + "***" + domain

df["Masked_Email"] = df["Email"].apply(mask_email)

print(df[["Email", "Masked_Email"]])
```

```python
def age_group(age):
    if 20 <= age <= 25:
        return "20-25"
    elif 26 <= age <= 30:
        return "26-30"
    elif 31 <= age <= 35:
        return "31-35"
    elif 36 <= age <= 40:
        return "36-40"

df["Age_Group"] = df["Age"].apply(age_group)

print(df[["Age", "Age_Group"]])
```

```python
region_map = {
    "Ahmedabad": "Gujarat",
    "Surat": "Gujarat",
    "Rajkot": "Gujarat",
    "Vadodara": "Gujarat",
    "Gandhinagar": "Gujarat",

    "Mumbai": "Maharashtra",
    "Pune": "Maharashtra",
    "Nashik": "Maharashtra",
    "Nagpur": "Maharashtra",
    "Bhopal": "Maharashtra",

    "Delhi": "Other India",
    "Kochi": "Other India",
    "Jaipur": "Other India",
    "Chennai": "Other India",
    "Hyderabad": "Other India",
    "Noida": "Other India",
    "Bengaluru": "Other India",
    "Indore": "Other India",
    "Lucknow": "Other India",
    "Thane": "Other India"
}

df["Region"] = df["City"].map(region_map)

print(df[["City", "Region"]])
```

```python
df["Anonymous_ID"] = [
    f"Customer_{i:03d}"
    for i in range(1, len(df) + 1)
]

print(df[["Name", "Anonymous_ID"]])
```

```python
protected_df = df[
    [
        "Customer_ID",
        "Anonymous_ID",
        "Age_Group",
        "Region",
        "Masked_Email",
        "Purchase_Amount",
        "Payment_Method"
    ]
].copy()

print("Privacy-Preserved Dataset:")
display(protected_df)
```
